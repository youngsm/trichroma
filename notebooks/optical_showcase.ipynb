### Try this

1. **Split white light.** Follow the transmitted rainbow and the reflected branches.
   Switch polarization and rerun to change the Fresnel probabilities.
2. **Turn ultraviolet into visible light.** Violet marks the otherwise invisible UV beam.
   The coating chooses a new visible wavelength and a delay from its actual input tables.
   Uncheck **Show all times** and advance **Time** to reveal the delayed emission.
3. **Scatter blue, transmit red.** Blue light has a shorter mean free path. The plot compares
   the measured fraction scattered at least once with the exponential free-flight law.

Click **Trace the light** after changing the experiment, count, seed or polarization.
Time controls only redraw the saved event. They reveal completed interaction segments,
not an interpolation of a photon's waiting time into its speed.

The optics are explicitly **synthetic**. These are projected Monte Carlo paths, not a
camera-radiance or brightness prediction. UV has a false display color and the line glow
is an illustration effect. The fluorescence example uses the implemented **surface**
reemission model; it does not introduce bulk reemission.
